**ATIVIDADE DE FIXAÇÃO DICONARIOS EM PYTHON**

Criar um dicionário chamado produto com as chaves: nome, preco e estoque. Depois imprimir todas as informações.

```python
produto = {
    "nome": "Notebook",
    "preco": 3500.00,
    "estoque": 10
}

print("Nome:", produto["nome"])
print("Preço:", produto["preco"])
print("Estoque:", produto["estoque"])
```

Criar um dicionário com três pessoas e suas idades. Mostrar apenas os nomes.

```python
pessoas = {
    "Ana": 25,
    "Carlos": 30,
    "Mariana": 22
}

for nome in pessoas.keys():
    print(nome)
```

Criar um dicionário com notas de um aluno em três disciplinas. Calcular e mostrar a média.

```python
notas = {
    "Matemática": 8.0,
    "Português": 7.5,
    "História": 9.0
}

media = sum(notas.values()) / len(notas)

print("Média:", media)
```

Cadastrar vários alunos (nome e média) usando dicionários dentro de uma lista. Mostrar apenas os aprovados (média ≥ 7).

```python
alunos = [
    {"nome": "João", "media": 8.5},
    {"nome": "Maria", "media": 6.0},
    {"nome": "Pedro", "media": 7.2},
    {"nome": "Ana", "media": 5.5}
]

print("Alunos aprovados:")

for aluno in alunos:
    if aluno["media"] >= 7:
        print(aluno["nome"], "- Média:", aluno["media"])
```
